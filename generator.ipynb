```python
import pandas as pd

dataset = pd.read_csv("sentences.csv.zip", compression="zip" )

train_dataset = dataset.sample(frac=0.8, random_state=0)
test_dataset = dataset.drop(train_dataset.index)

for dir in ["train", "test"]:
    dataset = train_dataset if dir == "train" else test_dataset
    for target in ["positive", "negative", "neutral"]:
        df = dataset[dataset["target"] == target].reset_index(drop=True)
        for index, row in df.iterrows():
            with open(f"{dir}/{target}/{index:04d}.txt", "w") as f:
                f.write(row["phrase"])
    
```
